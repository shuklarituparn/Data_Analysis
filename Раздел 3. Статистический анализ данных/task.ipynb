```python
# Оцените среднее время ответа веб-сервера с расчётом доверительного интервала.

# Предположим, у нас есть логи времени ответа веб-сервера на запросы, и мы хотим оценить среднее время ответа этого веб-сервера с учётом неопределённости. Для этого мы можем рассчитать доверительный интервал для среднего времени ответа на основе выборочных данных.

# Выборка с временем ответа веб-сервера:
# response_times = [150, 130, 140, 160, 180, 170, 190, 200, 175, 165]

# Чему равна левая граница доверительного интервала при уровне доверия 95% с точностью до двух знаков после запятой?


import numpy as np
import scipy.stats as stats

# Given response times
response_times = np.array([150, 130, 140, 160, 180, 170, 190, 200, 175, 165])

# Sample mean and standard error
mean_response = np.mean(response_times)
std_error = stats.sem(response_times)

# 95% confidence interval
confidence = 0.95
t_value = stats.t.ppf((1 + confidence) / 2, df=len(response_times) - 1)

# Confidence interval bounds
lower_bound = mean_response - t_value * std_error
upper_bound = mean_response + t_value * std_error

# Print the left bound rounded to 2 decimal places
print(f"Left bound of 95% confidence interval: {lower_bound:.2f}")
```

```python
from scipy.stats import mannwhitneyu

# Given response times for Algorithm A and Algorithm B
algorithm_a_times = [3.2, 5.1, 4.3, 6.7, 7.9, 4.8, 5.6, 6.2, 5.9]
algorithm_b_times = [2.9, 4.5, 6.1, 5.8, 4.2, 5.3, 4.7, 6.5, 6.8]

# Perform the Mann-Whitney U test
statistic, p_value = mannwhitneyu(algorithm_a_times, algorithm_b_times, alternative='two-sided')

# Print the result rounded to one decimal place
print(f"Statistic: {round(statistic, 1)}")
```
